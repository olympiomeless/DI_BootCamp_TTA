## Exercise 2: Multi-Part Prompt for Quiz Generation
**Scenario**: Create an educational prompt for 7th graders about volcanic eruptions.

### Tasks
- Ask for: 2-bullet summary + 3 MCQs (1 correct, 2 distractors).
- Require friendly tone, simple vocab (11?13 y/o), and Google Slides-ready output.
- Explain 3 ways your prompt improves on: `Make a quiz for kids about this article.`

```python
quiz_prompt = """You are an engaging 7th-grade science teacher. Your task is to create educational content about volcanic eruptions for a Google Slides presentation. Based on the provided article about volcanic eruptions (ARTICLE_TEXT_HERE), generate the following:

1.  **Summary:** Provide a 2-bullet point summary using simple vocabulary suitable for 11-13 year olds.
2.  **Multiple Choice Questions:** Generate 3 multiple-choice questions about the article. Each question should have one correct answer and two plausible, but incorrect, distractors. Ensure the language is friendly and easy for 7th graders to understand.

**Format:** Present the summary first, followed by each question. For the multiple-choice questions, list the question, then the three options (A, B, C), and finally, clearly indicate the correct answer. The overall tone should be enthusiastic and encouraging."""
quiz_prompt
```

```python
improvements = [
    'Specific Output Structure: My prompt explicitly asks for a 2-bullet summary and 3 MCQs with 1 correct and 2 distractors, preventing the model from generating random content or an unstructured quiz.',
    'Defined Audience and Tone: By specifying "7th-grade science teacher," "simple vocabulary for 11-13 year olds," and "friendly tone," the prompt ensures the output is age-appropriate and engaging, unlike a generic "quiz for kids."',
    'Google Slides-ready Format: Requiring a specific output format (summary first, then questions with options and correct answers) ensures the content is immediately usable and reduces post-generation editing.'
]
improvements
```

## Exercise 6: Hallucination Spotting and Mitigation
**Scenario**: Model added an unsupported claim in a climate/marine article summary.

### Tasks
1) Prompt the model to avoid hallucinations using only provided text.
2) Rewrite with verification language.
3) List 2 mitigation strategies.
4) Name 2 domains where hallucinations are risky and why.

```python
no_hallucination_prompt = """Summarize the following article (ARTICLE_TEXT_HERE) about climate change and marine biodiversity. Base your summary STRICTLY and ONLY on the information explicitly provided in the article. Do not introduce any outside facts, statistics, or claims. If a piece of information is not present in the article, do not include it in your summary."""
no_hallucination_prompt
```

```python
verification_prompt = """Summarize the following article (ARTICLE_TEXT_HERE) about climate change and marine biodiversity. Only include claims and data explicitly stated and verifiable within the provided text. If you are uncertain about the origin of a piece of information or if it cannot be directly attributed to the article, state that the information is not available in the provided text rather than generating a potentially incorrect claim. Every statement must be grounded in the article."""
verification_prompt
```

```python
mitigation_strategies = [
    'Grounding to Source Material: Explicitly instructing the model to use ONLY the provided text and to not introduce outside information.',
    'Verification Language: Using phrases like "Only include claims explicitly stated" and "If uncertain, state that the information is not available" to guide the model towards caution and transparency.'
]
risky_domains = {
    'domain_1': {
        'name': 'Healthcare/Medical Diagnosis',
        'why': 'Hallucinations in this domain could lead to incorrect diagnoses, inappropriate treatments, or false health information, directly endangering patient lives and well-being.'
    },
    'domain_2': {
        'name': 'Legal Advice',
        'why': 'Generating hallucinated legal precedents, statutes, or advice could result in misinformed legal strategies, incorrect filings, or severe legal repercussions for individuals or organizations.'
    }
}
risky_domains
```
