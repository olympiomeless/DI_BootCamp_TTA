### My answer

For a binary loan default prediction task, the two most suitable candidate models are **Logistic Regression** and **Random Forest Classifier**. The evaluation plan begins with a stratified train/test split (80/20) to preserve the class ratio, followed by 5-fold stratified cross-validation on the training set to obtain reliable performance estimates that are not sensitive to a single split. The metrics reported will be **Precision**, **Recall**, **F1-score**, **ROC-AUC**, and **PR-AUC (Average Precision)**, because accuracy alone is misleading on the imbalanced class distribution (≈69% approved vs. ≈31% default). ROC-AUC measures the model's ability to discriminate between classes across all thresholds, while PR-AUC is more informative when the positive class (default) is the minority. A confusion matrix will be computed on the held-out test set to quantify false negatives (missed defaults, which are costly for the institution) and false positives (wrongly denied loans, which hurt applicants).

The dataset is moderately imbalanced (422 approved vs. 192 defaults). To address this, we will use `class_weight='balanced'` in Logistic Regression and Random Forest, which adjusts the loss function to penalize misclassification of the minority class proportionally; alternatively, SMOTE oversampling can be applied to the training fold only within each cross-validation split to avoid data leakage.

The decision threshold will be set based on the business cost of false negatives versus false positives: since missing a default is typically more costly than denying a loan, the threshold will be tuned on the precision-recall curve to achieve a Recall ≥ 0.80 on the default class, even at the cost of some precision. **Scenario 1 — Predicting Stock Prices (Supervised Learning: Regression)**

Predicting future stock prices is best addressed using **supervised learning**, specifically a regression approach, because historical labeled data (past prices and associated features) is available and the output is a continuous numeric value. The input data consists of historical time series of prices, trading volumes, technical indicators (moving averages, RSI), and potentially macroeconomic features. The output is a future price or return value at a specified horizon (e.g., the closing price tomorrow). The learning signal is the minimization of a continuous loss function such as Mean Squared Error (MSE) or Mean Absolute Error (MAE) between the predicted and actual prices. Suitable models include LSTM networks or gradient-boosted trees (XGBoost, LightGBM) for tabular feature-based prediction.

**Scenario 2 — Organizing a Library of Books (Unsupervised Learning: Clustering)**

Grouping books by genre or category based on content similarity is best addressed using **unsupervised learning**, specifically a clustering algorithm, because there are no predefined labeled categories — the model must discover structure autonomously from the data. The input data consists of numeric representations of book content, such as TF-IDF vectors or semantic embeddings (e.g., from a sentence transformer) derived from titles, descriptions, and full text. The output is a set of cluster assignments, each representing a discovered genre or thematic group. The learning objective is to maximize intra-cluster similarity and inter-cluster separation, measured by metrics such as the silhouette score; K-Means or hierarchical clustering are appropriate algorithms, with the optimal number of clusters determined by the elbow method or silhouette analysis.

**Scenario 3 — Robot Navigating a Maze (Reinforcement Learning)**

Programming a robot to find the shortest path in a maze is best addressed using **reinforcement learning (RL)**, because the robot must learn an optimal policy through trial-and-error interaction with its environment without a pre-labeled dataset of correct moves. The input data is the robot's current state (its position in the maze, sensor readings), and the output is an action (move up, down, left, or right). The learning signal is a reward function: the robot receives a positive reward for reaching the exit and a small negative reward (penalty) for each step taken, which incentivizes it to find the shortest path; algorithms such as Q-Learning or Deep Q-Network (DQN) are well-suited for this discrete action-space navigation task.

### My answer

**Supervised Learning — Classification Model (e.g., Random Forest for Loan Default Prediction)**

For a supervised classification model, the evaluation strategy begins with a stratified train/validation/test split (60/20/20) to ensure each subset reflects the class distribution. A 5-fold stratified cross-validation is then applied on the training set to obtain stable performance estimates across multiple data splits. The key metrics reported are Accuracy, Precision, Recall, F1-score, ROC-AUC, and PR-AUC (Average Precision): Accuracy gives the overall correct classification rate but is misleading under class imbalance; Precision measures how many predicted positives are truly positive; Recall (Sensitivity) measures how many actual positives are correctly identified; F1-score balances Precision and Recall; ROC-AUC quantifies the model's ranking ability across all thresholds; and PR-AUC is particularly informative for imbalanced datasets where the minority class is of primary interest. The ROC curve and Precision-Recall curve are plotted on the held-out test set. The decision threshold is selected by finding the point on the PR curve that satisfies a minimum Recall constraint defined by the business cost of false negatives. A key evaluation challenge for supervised models is data leakage: if any preprocessing step (imputation, scaling, feature selection) is fitted on the full dataset before the train/test split, evaluation metrics will be optimistically biased and will not reflect true generalization performance.

**Unsupervised Learning — Clustering Model (e.g., K-Means for Book Categorization)**

For an unsupervised clustering model, evaluation cannot rely on ground-truth labels. Cluster quality is assessed using internal validation metrics: the **Silhouette Score** measures how similar each point is to its own cluster compared to other clusters, with values ranging from -1 (poor) to +1 (ideal), and a score above 0.5 generally indicates well-separated, compact clusters. The **Elbow Method** is used to select the optimal number of clusters K: inertia (within-cluster sum of squares) is plotted as a function of K, and the "elbow" — the point at which the rate of decrease sharply flattens — indicates the best trade-off between model complexity and cluster compactness. The **Davies-Bouldin Index** provides a complementary measure where lower values indicate better separation. A t-SNE or PCA 2D projection of the clustered data is plotted to visually inspect whether the clusters are coherent and well-separated. The primary evaluation challenge for unsupervised models is the absence of ground truth: all internal metrics measure geometric properties of the clusters (compactness and separation) rather than semantic correctness, so a high silhouette score does not guarantee that the discovered clusters align with meaningful real-world categories.

**Reinforcement Learning — Q-Learning Agent (e.g., Maze Navigation Robot)**

For a reinforcement learning model, success is measured primarily by the **cumulative reward** per episode: a learning curve of the total reward obtained in each episode is plotted over training; a consistently increasing trend that eventually plateaus indicates that the agent is learning an increasingly optimal policy. **Convergence** is assessed by monitoring the stability of the Q-values or policy over successive episodes — if the policy stops changing significantly between updates, the agent has converged. The **Exploration vs. Exploitation balance** is managed through an epsilon-greedy strategy in which epsilon (the probability of taking a random action) decays over time: early in training, high epsilon encourages exploration of the state space, while later in training, low epsilon allows the agent to exploit the learned optimal policy. Additional metrics include the average number of steps per episode to reach the goal (which should decrease toward the optimal shortest path) and the success rate across multiple test episodes. The main evaluation challenge for RL models is the high variance inherent in the learning process: results are sensitive to the choice of hyperparameters (learning rate, discount factor γ, epsilon decay schedule), the random seed, and the reward function design; a poorly shaped reward can lead to suboptimal or degenerate policies even when cumulative reward appears to improve.
