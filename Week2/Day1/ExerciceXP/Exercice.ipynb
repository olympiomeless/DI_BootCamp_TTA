## Exercice 1

**1.Qu'est ce que l'analyse de données**

L'analyse de données est un processus d' inspection , de nettoyage , de transformation et de modélisation des données visant à en extraire des informations utiles, à éclairer les conclusions et à faciliter la prise de décision

**2.Pourquoi l'analyse des données est-elle importante ?**

Dans les contextes modernes, l'analyse de données est devenue indispensable pour plusieurs raisons :

- Prise de décision éclairée : Les organisations s'appuient sur des données concrètes plutôt que sur l'intuition, réduisant les risques d'erreur.
- Compétitivité : Les entreprises qui exploitent leurs données prennent de l'avance en identifiant les tendances du marché avant leurs concurrents.
- Optimisation des processus : L'analyse permet de détecter les inefficacités et d'améliorer les opérations.
Personnalisation : Les services peuvent être adaptés aux besoins spécifiques des utilisateurs grâce à la compréhension de leurs comportements.

**3.Trois domaines d'application actuels de l'analyse de données**
- Santé
- Marketing
- Climat

```python
def charger_dataset(chemin_fichier, n_lignes=5):

    try:
        if chemin_fichier.endswith('.csv'):
            df = pd.read_csv(chemin_fichier)

        elif chemin_fichier.endswith(('.xlsx', '.xls')):
            df = pd.read_excel(chemin_fichier)

        else:
            print("Unsupported file format.")
            return

        print("\n=== First lines of the dataset ===")
        print(df.head(n_lignes))

        
        print("\n=== General Information ===")
        print(f"Number of rows : {df.shape[0]}")
        print(f"Number of columns : {df.shape[1]}")

        print("\n=== Column types ===")
        print(df.dtypes)

        print("\n=== Missing Values ===")
        print(df.isnull().sum())

        print("\n=== Statistical Description ===")
        print(df.describe(include='all'))

        return df

    except FileNotFoundError:
        print("Not Found file.")
    
    except Exception as e:
        print(f"Erreur : {e}")

# Exemple d'utilisation
chemin = 'C:\\Users\\meles\\Documents\\TTA_DI_BootCamp_Gilles-Chris_MAKE\\Week2\\Day1\\ExerciceXP\\Dataset\\Mental health Depression disorder Data.csv'
chemin1 = 'C:\\Users\\meles\\Documents\\TTA_DI_BootCamp_Gilles-Chris_MAKE\\Week2\\Day1\\ExerciceXP\\Dataset\\clean_dataset.csv'
chemin2 = 'C:\\Users\\meles\\Documents\\TTA_DI_BootCamp_Gilles-Chris_MAKE\\Week2\\Day1\\ExerciceXP\\Dataset\\crx.csv'
chemin3 = 'C:\\Users\\meles\\Documents\\TTA_DI_BootCamp_Gilles-Chris_MAKE\\Week2\\Day1\\ExerciceXP\\Dataset\\Time Americans Spend Sleeping.csv'
dk = charger_dataset(chemin)
# dg = charger_dataset(chemin1)
# dh = charger_dataset(chemin2)
# di = charger_dataset(chemin3)
```

```python
transformations = [
    {
        'Source': '📝 Articles de blog — expériences de voyage',
        'Méthode': 'NLP (Traitement du Langage Naturel) + Extraction d\'entités nommées (NER)',
        'Colonnes résultantes': ['Auteur', 'Date', 'Pays visité', 'Ville', 'Type de voyage',
                                  'Note globale', 'Thèmes principaux', 'Sentiments (positif/négatif)'],
        'Justification': 'Le NER extrait automatiquement les lieux, dates et noms. L\'analyse de sentiment quantifie l\'expérience subjective.'
    },
    {
        'Source': '🎙️ Enregistrements audio — appels au service client',
        'Méthode': 'Transcription automatique (ASR) + Classification de texte',
        'Colonnes résultantes': ['ID_appel', 'Durée (s)', 'Agent ID', 'Problème catégorisé',
                                  'Sentiment client', 'Résolution (Oui/Non)', 'Mots-clés'],
        'Justification': 'Des outils comme Whisper (OpenAI) transcrivent l\'audio. Le texte est ensuite classifié par type de problème et sentiment.'
    },
    {
        'Source': '✍️ Notes manuscrites — séance de brainstorming',
        'Méthode': 'OCR (Reconnaissance Optique de Caractères) + Regroupement thématique',
        'Colonnes résultantes': ['ID_idée', 'Texte extrait', 'Auteur', 'Thème', 'Priorité', 'Date'],
        'Justification': 'L\'OCR (ex: Tesseract) convertit les images en texte. Un modèle de clustering groupe ensuite les idées similaires.'
    },
    {
        'Source': '🎥 Tutoriel vidéo — cuisine',
        'Méthode': 'Transcription audio + Vision par ordinateur + Extraction de recettes',
        'Colonnes résultantes': ['Titre_recette', 'Ingrédients', 'Quantités', 'Étapes', 'Durée (min)',
                                  'Difficulté', 'Nombre de portions'],
        'Justification': 'La transcription audio fournit les instructions. La vision par ordinateur identifie les ingrédients visuellement. Un modèle LLM structure le tout en recette.'
    }
]

for t in transformations:
    print(f"\n{'='*60}")
    print(f"Source : {t['Source']}")
    print(f"Method : {t['Méthode']}")
    print(f"Colonnes résultantes : {', '.join(t['Colonnes résultantes'])}")
    print(f"Justification : {t['Justification']}")
```
