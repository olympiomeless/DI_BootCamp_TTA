```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
wheather_array = np.random.uniform(low=-5, high=35, size=(10, 12))

villes = [
    "Abidjan", "Paris", "Londres", "Tokyo", "New York",
    "Berlin", "Madrid", "Dakar", "Cotonou", "Nairobi"
]

mois = [
    "Jan", "Fév", "Mar", "Avr", "Mai", "Juin",
    "Juil", "Août", "Sep", "Oct", "Nov", "Déc"
]

wheather_data = pd.DataFrame(wheather_array, index=villes, columns=mois)
print(wheather_data)
```

```python
sum_temp = wheather_data.mean(axis=1)
print("Température moyenne (°C) :")
print(sum_temp.round(2))
```

```python
hottest_city = sum_temp.idxmax()
print(f"\nLa ville la plus chaude en moyenne est : {hottest_city} avec une température moyenne de {sum_temp[hottest_city]:.2f} °C.")
ice_cold_city = sum_temp.idxmin()
print(f"La ville la plus froide en moyenne est : {ice_cold_city} avec une température moyenne de {sum_temp[ice_cold_city]:.2f} °C.")
```

```python
fig, ax = plt.subplots(figsize=(10, 6))
wheather_data.T.plot(kind='bar', ax=ax)
plt.xlabel('Mois')
plt.ylabel('Température (°C)')
plt.title('Température moyenne par mois')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```
