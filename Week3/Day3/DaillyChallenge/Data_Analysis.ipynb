## 1. Chargement des données (Data Loading)

```python
import pandas as pd

# Load the training dataset
try:
    df_train = pd.read_csv('train.csv')
    print("Train data loaded successfully.")
except FileNotFoundError:
    print("Error: 'train.csv' not found. Please ensure the file is in the correct directory.")
    df_train = None

# Load the test dataset (for later use)
try:
    df_test = pd.read_csv('test.csv')
    print("Test data loaded successfully.")
except FileNotFoundError:
    print("Error: 'test.csv' not found. Please ensure the file is in the correct directory.")
    df_test = None
```

## 2. Exploration de l'ensemble de données d'entraînement (Exploring the Training Dataset)

```python
if df_train is not None:
    print("\nShape of the training data:", df_train.shape)
    print("\nFirst 5 rows of the training data:")
    display(df_train.head())

    print("\nInformation about the training data (data types, non-null counts):")
    display(df_train.info())

    print("\nBasic descriptive statistics for the training data:")
    display(df_train.describe())
```

## 3. Analyse statistique avec NumPy et SciPy

### Mesures de Tendance Centrale, Variabilité et Forme de Distribution (Measures of Central Tendency, Variability, and Distribution Shape)

```python
import numpy as np
from scipy.stats import skew, kurtosis

if df_train is not None:
    print("\nStatistical Analysis for each Feature:\n")
    stats_df = pd.DataFrame(columns=['Mean', 'Median', 'Mode', 'Range', 'Variance', 'Std Dev', 'Skewness', 'Kurtosis'])

    # Identify numerical columns (excluding the target 'price_range' for individual stats, but including it in the loop for completeness)
    numerical_cols = df_train.select_dtypes(include=np.number).columns.tolist()

    for col in numerical_cols:
        # Mean and Median
        mean_val = df_train[col].mean()
        median_val = df_train[col].median()

        # Mode (can return multiple values, so take the first if multiple or NaN if none)
        mode_val_series = df_train[col].mode()
        mode_val = mode_val_series.iloc[0] if not mode_val_series.empty else np.nan

        # Range
        range_val = df_train[col].max() - df_train[col].min()

        # Variance and Standard Deviation
        variance_val = df_train[col].var()
        std_dev_val = df_train[col].std()

        # Skewness and Kurtosis
        skewness_val = skew(df_train[col])
        kurtosis_val = kurtosis(df_train[col])

        stats_df.loc[col] = [mean_val, median_val, mode_val, range_val, variance_val, std_dev_val, skewness_val, kurtosis_val]
    
    print("Detailed Statistical Measures for each Feature:")
    display(stats_df.round(3))

    print("\nSummary of target variable 'price_range':")
    display(df_train['price_range'].value_counts().sort_index())
```

### Tests d'hypothèse et Analyse de Corrélation (Hypothesis Testing and Correlation Analysis)

#### Test ANOVA pour la RAM par gamme de prix (ANOVA Test for RAM by Price Range)

```python
from scipy.stats import f_oneway

if df_train is not None:
    # Group RAM by price_range
    price_groups = []
    for i in sorted(df_train['price_range'].unique()):
        price_groups.append(df_train[df_train['price_range'] == i]['ram'])

    # Perform one-way ANOVA
    f_statistic, p_value = f_oneway(*price_groups)

    print(f"ANOVA Test for RAM across Price Ranges:")
    print(f"F-statistic: {f_statistic:.3f}")
    print(f"P-value: {p_value:.3f}")

    if p_value < 0.05:
        print("Conclusion: The p-value is less than 0.05, indicating a statistically significant difference in mean RAM across different price ranges.")
    else:
        print("Conclusion: The p-value is not less than 0.05, indicating no statistically significant difference in mean RAM across different price ranges.")
```

#### Matrice de Corrélation (Correlation Matrix)

```python
if df_train is not None:
    # Calculate the correlation matrix
    correlation_matrix = df_train.corr(method='pearson')

    # Display correlations with the target variable 'price_range'
    print("\nCorrelation of Features with 'price_range':")
    display(correlation_matrix['price_range'].sort_values(ascending=False))
```

## 4. Visualisation des données avec Matplotlib (Data Visualization with Matplotlib)

### Histograms for Key Features

```python
import matplotlib.pyplot as plt
import seaborn as sns

if df_train is not None:
    # Select a few key numerical features for histograms
    features_to_plot = ['ram', 'battery_power', 'px_width', 'px_height', 'int_memory']

    plt.figure(figsize=(15, 10))
    for i, feature in enumerate(features_to_plot):
        plt.subplot(2, 3, i + 1) # Adjust subplot grid based on number of features
        sns.histplot(df_train[feature], kde=True)
        plt.title(f'Distribution of {feature}')
        plt.xlabel(feature)
        plt.ylabel('Frequency')
    plt.tight_layout()
    plt.show()
```

### Box Plots for Features vs. Price Range

```python
if df_train is not None:
    # Select features that showed significant correlation or are important
    features_for_boxplot = ['ram', 'battery_power', 'px_width', 'px_height']

    plt.figure(figsize=(15, 10))
    for i, feature in enumerate(features_for_boxplot):
        plt.subplot(2, 2, i + 1)
        sns.boxplot(x='price_range', y=feature, data=df_train)
        plt.title(f'{feature} by Price Range')
        plt.xlabel('Price Range')
        plt.ylabel(feature)
    plt.tight_layout()
    plt.show()
```

## 5. Synthèse des idées et conclusion (Synthesis of Insights and Conclusion)

Based on the statistical analysis and visualizations, we can draw the following conclusions:

**Key Determinants of Mobile Phone Price Classification:**
1.  **RAM (Random Access Memory):** This is by far the most significant factor influencing the price range. The ANOVA test showed a highly significant difference in RAM across price ranges, and its correlation with `price_range` is extremely strong (0.917).
2.  **Battery Power:** There is a moderate positive correlation (0.201) between battery power and price range. Phones with higher battery capacity tend to be in higher price brackets.
3.  **Pixel Resolution (px_width, px_height):** Both screen width and height in pixels show a positive, albeit weaker, correlation with price range (0.166 and 0.149 respectively). This suggests that phones with higher resolution screens are generally more expensive.

**Other Observations:**
*   **Internal Memory (int_memory):** Has a weak positive correlation (0.044).
*   **Camera Features (fc, pc):** Front camera (fc) and primary camera (pc) megapixels have very weak positive correlations.
*   **Connectivity (three_g, four_g, wifi, blue):** These binary features show very weak positive correlations, suggesting that while their presence is expected, they are not strong differentiators for price once available.
*   **Physical Attributes (mobile_wt, sc_h, sc_w, m_dep):** Mobile weight, screen height, screen width (physical), and depth (m_dep) show negligible to very weak correlations, indicating they are not primary drivers of price.
*   **Clock Speed (clock_speed), Number of Cores (n_cores), Dual SIM (dual_sim), Touch Screen (touch_screen):** These features also have very weak or near-zero correlations with price range.

**Unexpected or Significant Discoveries:**
*   The overwhelming dominance of **RAM** as a price predictor is a crucial insight. It suggests that consumers are willing to pay significantly more for higher RAM capacities.
*   Many features that one might intuitively consider important (like `n_cores`, `clock_speed`, `mobile_wt`) show very little linear correlation with the `price_range` in this dataset. This indicates that their impact on price might be indirect, non-linear, or overshadowed by other features, especially RAM.
*   The relatively low correlation of `battery_power` compared to `ram` is also noteworthy.

In summary, for this dataset, **RAM is the primary differentiator for mobile phone price ranges**, followed by battery power and screen resolution. Other features play a much smaller, if any, role in determining the price category.
