## A. Server (server.py)
Create a small MCP server named "Demo" with:
- Tool `add(a: int, b: int) -> int` returning the sum.
- Resource template `greeting://{name}` returning "Hello, {name}!".
- Start the STDIO loop in `__main__`.

````python
%%writefile server.py
from mcp.server.fastmcp import FastMCP

mcp = FastMCP("Demo")

@mcp.tool()
def add(a: int, b: int) -> int:
    """Return the sum of two integers."""
    return a + b

@mcp.tool()
def subtract(a: int, b: int) -> int:
    """Return the difference of two integers."""
    return a - b

@mcp.resource("greeting://{name}")
def greet(name: str) -> str:
    """Return a greeting for the given name."""
    return f"Hello, {name}!"

if __name__ == "__main__":
    mcp.run_stdio_server()
````

## B. Client (client.py)
Write a client that:
1) Spawns the server via STDIO using the MCP CLI.
2) Initializes a session.
3) Lists resources and tools, printing their names.
4) Reads `greeting://hello` and prints it.
5) Calls tool `add` with a=1, b=7 and prints the result.

````python
%%writefile client.py
import asyncio
from mcp import ClientSession, StdioServerParameters
from mcp.client.stdio import stdio_client

server_params = StdioServerParameters(command="mcp", args=["run", "server.py"], env=None)


def extract_content(payload):
    """Best-effort to pull text from MCP responses."""
    if hasattr(payload, "contents"):
        contents = payload.contents
        if contents:
            first = contents[0]
            if hasattr(first, "text"):
                return first.text
            if isinstance(first, dict) and "text" in first:
                return first["text"]
            return str(first)
    if hasattr(payload, "content"):
        return payload.content
    return str(payload)


async def run():
    async with stdio_client(server_params) as (read, write):
        async with ClientSession(read, write) as session:
            await session.initialize()

            # List resources and print their URIs
            list_resources_response = await session.list_resources()
            print("Resources:")
            if list_resources_response.resources:
                for resource in list_resources_response.resources:
                    print(f"  - {resource.uri}")
            else:
                print("  No resources found.")

            # List tools and print their names
            list_tools_response = await session.list_tools()
            print("Tools:")
            if list_tools_response.tools:
                for tool in list_tools_response.tools:
                    print(f"  - {tool.name}")
            else:
                print("  No tools found.")

            # Read greeting://hello and print the content
            greeting_payload = await session.read_resource("greeting://hello")
            print(f"Greeting: {extract_content(greeting_payload)}")

            # Call add with a=1, b=7 and print the result
            add_result_payload = await session.call_tool("add", {"a": 1, "b": 7})
            print(f"Add result (1 + 7): {extract_content(add_result_payload)}")

            # Call subtract with a=10, b=3 and print the result
            subtract_result_payload = await session.call_tool("subtract", {"a": 10, "b": 3})
            print(f"Subtract result (10 - 3): {extract_content(subtract_result_payload)}")


if __name__ == "__main__":
    asyncio.run(run())
````
