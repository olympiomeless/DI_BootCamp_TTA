## Chargement et Inspection des Données MNIST

```python
import keras
from keras.datasets import mnist

# Charger l'ensemble de données MNIST
(x_train, y_train), (x_test, y_test) = mnist.load_data()

# Imprimer les formes des données chargées
print(f"Forme de x_train: {x_train.shape}")
print(f"Forme de y_train: {y_train.shape}")
print(f"Forme de x_test: {x_test.shape}")
print(f"Forme de y_test: {y_test.shape}")
```

## Prétraitement des Données pour un Réseau Neuronal Entièrement Connecté

```python
import numpy as np
from keras.utils import to_categorical

# Aplatir les images de 28x28 à 784 pixels
x_train_flattened = x_train.reshape((x_train.shape[0], 28 * 28))
x_test_flattened = x_test.reshape((x_test.shape[0], 28 * 28))

# Normaliser les valeurs des pixels entre 0 et 1
x_train_normalized = x_train_flattened.astype('float32') / 255
x_test_normalized = x_test_flattened.astype('float32') / 255

# Encoder les étiquettes cibles en one-hot
y_train_one_hot = to_categorical(y_train, num_classes=10)
y_test_one_hot = to_categorical(y_test, num_classes=10)

print(f"Forme de x_train_normalized: {x_train_normalized.shape}")
print(f"Forme de y_train_one_hot: {y_train_one_hot.shape}")
print(f"Forme de x_test_normalized: {x_test_normalized.shape}")
print(f"Forme de y_test_one_hot: {y_test_one_hot.shape}")
```

## Construire et Entraîner un Réseau Neuronal Entièrement Connecté (Fully Connected Neural Network)

```python
from keras.models import Sequential
from keras.layers import Dense

# Créer un modèle Sequential
model_fc = Sequential([
    Dense(256, activation='relu', input_shape=(784,)),
    Dense(128, activation='relu'),
    Dense(10, activation='softmax')
])

# Compiler le modèle
model_fc.compile(optimizer='adam',
              loss='categorical_crossentropy',
              metrics=['accuracy'])

# Afficher le résumé du modèle
model_fc.summary()

# Entraîner le modèle
history_fc = model_fc.fit(x_train_normalized, y_train_one_hot, epochs=10, batch_size=128, validation_split=0.2)

# Évaluer les performances du modèle
loss_fc, accuracy_fc = model_fc.evaluate(x_test_normalized, y_test_one_hot, verbose=0)
print(f"Précision du modèle FC sur l'ensemble de test: {accuracy_fc:.4f}")
```

## Prétraitement des Données pour un Réseau Neuronal Convolutif (CNN)

```python
import numpy as np
from keras.utils import to_categorical

# Remodeler les données d'entrée à la forme attendue par une couche Conv2D (par exemple, (60000, 28, 28, 1))
# x_train et x_test sont déjà de forme (images, hauteur, largeur), nous devons ajouter une dimension pour les canaux (gris).
x_train_reshaped = x_train.reshape(x_train.shape[0], 28, 28, 1)
x_test_reshaped = x_test.reshape(x_test.shape[0], 28, 28, 1)

# Normaliser les valeurs des pixels en les divisant par 255
x_train_cnn_normalized = x_train_reshaped.astype('float32') / 255
x_test_cnn_normalized = x_test_reshaped.astype('float32') / 255

# Encoder les étiquettes cibles en one-hot (si ce n'est pas déjà fait pour le CNN, nous utilisons les y_train, y_test originaux)
y_train_cnn_one_hot = to_categorical(y_train, num_classes=10)
y_test_cnn_one_hot = to_categorical(y_test, num_classes=10)

print(f"Forme de x_train_cnn_normalized: {x_train_cnn_normalized.shape}")
print(f"Forme de y_train_cnn_one_hot: {y_train_cnn_one_hot.shape}")
print(f"Forme de x_test_cnn_normalized: {x_test_cnn_normalized.shape}")
print(f"Forme de y_test_cnn_one_hot: {y_test_cnn_one_hot.shape}")
```

## Concevoir et Entraîner un Réseau Neuronal Convolutif (CNN)

```python
from keras.models import Sequential
from keras.layers import Conv2D, MaxPooling2D, Flatten, Dense

# Créer un modèle Sequential
model_cnn = Sequential([
    Conv2D(32, (3, 3), activation='relu', input_shape=(28, 28, 1)),
    MaxPooling2D((2, 2)),
    Conv2D(64, (3, 3), activation='relu'),
    MaxPooling2D((2, 2)),
    Flatten(),
    Dense(128, activation='relu'),
    Dense(10, activation='softmax')
])

# Compiler le modèle
model_cnn.compile(optimizer='adam',
               loss='categorical_crossentropy',
               metrics=['accuracy'])

# Afficher le résumé du modèle
model_cnn.summary()

# Entraîner le modèle
history_cnn = model_cnn.fit(x_train_cnn_normalized, y_train_cnn_one_hot, epochs=10, batch_size=128, validation_split=0.2)

# Évaluer les performances du modèle
loss_cnn, accuracy_cnn = model_cnn.evaluate(x_test_cnn_normalized, y_test_cnn_one_hot, verbose=0)
print(f"Précision du modèle CNN sur l'ensemble de test: {accuracy_cnn:.4f}")
```

## Comparaison des Performances

```python
print(f"Précision finale du modèle entièrement connecté (FC) sur l'ensemble de test: {accuracy_fc:.4f}")
print(f"Précision finale du modèle convolutif (CNN) sur l'ensemble de test: {accuracy_cnn:.4f}")

if accuracy_cnn > accuracy_fc:
    print("Le modèle CNN a obtenu une meilleure précision que le modèle entièrement connecté.")
elif accuracy_fc > accuracy_cnn:
    print("Le modèle entièrement connecté a obtenu une meilleure précision que le modèle CNN.")
else:
    print("Les deux modèles ont obtenu la même précision.")
```
